**LangGraph runtime context -> `RuntimeContextSchema`**
- LangGraph + LangChain v1 기반의 '에이전트/그래프'를 본격적으로 사용한다면, 아래 방법을 추천한다.
- 상태 관리, 리소스 주입, 타입 안정성 측면에서 이점이 크다.

LangGraph 문서의 **runtime context** 개념을, 이 실습에서는 **`RuntimeContextSchema`**(Pydantic `BaseModel`)로 정의한다.

다만 `RuntimeContextSchema` 안에서 바로 비즈니스 로직을 전부 구현하지 말고,
- 핵심 로직은 별도 순수 함수/클래스로 분리
- tool 함수는 '컨텍스트에서 의존성 꺼내서 순수 함수를 호출'하는 역할만 하도록 만드는 것을 추천한다.

이렇게 하면
- LangGraph의 장점(컨텍스트, 체크포인팅, 타입)과
- 전통적인 소프트웨어 설계의 장점(테스트 용이성, 재사용성, 의존성 명시)을
- 둘 다 최대한 가져갈 수 있다.
- - -

- https://github.com/jinsooya/langgraph-runtime-context

```python
# --- stream: 'messages'
from langchain.messages import AIMessage

user_message = '각 직원이 담당하는 고객 수는 몇 명인가요?'
for state in agent.stream(
    input={'messages':[HumanMessage(user_message)]}, 
    context=RuntimeContextSchema(db=db, model=model),  # 런타임 컨텍스트 주입
    stream_mode='messages'
):
    msg, info = state            
    if isinstance(msg, AIMessage) and info.get('langgraph_node') == 'model':
        block = msg.content_blocks
        if block and block[0]['type'] != 'tool_call_state' and block[0]['type'] == 'text':
            print(block[0]['text'], end='', flush=True)
```

# 고객 세그멘테이션이란?


고객 세그멘테이션은 고객을 **유사한 특성끼리 그룹화**하는 분석 기법이다.

| 구분 | 내용 |
|------|------|
| **목적** | 고객을 유사한 집단으로 나눠 각 그룹에 맞는 마케팅 전략 수립 |
| **방법** | 비지도 학습(Unsupervised Learning) — 정답 없이 데이터 패턴만으로 그룹화 |
| **대표 알고리즘** | K-Means 군집화 |


## 왜 고객 세그멘테이션인가?


- **맞춤형 마케팅**: 
    + 모든 고객에게 동일한 메시지 -> 효율 낮음
    + 그룹별 맞춤 전략 -> 효율 높음
- **자원 집중**: 고가치 고객 그룹에 예산을 집중 투입할 수 있다.
- **신제품 개발**: 각 세그먼트의 소비 패턴으로 수요를 예측할 수 있다.


## K-Means 군집화 원리


K-Means는 데이터를 K개의 그룹으로 나누는 알고리즘이다.

| 단계 | 설명 |
|------|------|
| 1. 초기화 | K개의 중심점(centroid)을 임의로 설정 |
| 2. 할당 | 각 데이터 포인트를 가장 가까운 중심점에 배정 |
| 3. 업데이트 | 각 그룹의 평균으로 중심점 재계산 |
| 4. 반복 | 중심점이 변하지 않을 때까지 2~3단계 반복 |

**핵심 질문**: 몇 개의 그룹(K)으로 나눌 것인가? → **Elbow Method**로 결정한다.


## 이 실습에서의 군집화 피처


이 실습에서 사용할 마케팅 데이터의 군집화 피처는 다음과 같다.

| 피처 그룹 | 컬럼 | 의미 |
|----------|------|------|
| **소비 금액** | `MntWines`, `MntFruits`, `MntMeatProducts`, `MntFishProducts`, `MntSweetProducts`, `MntGoldProds` | 제품군별 총 소비금액 |
| **구매 채널** | `NumWebPurchases`, `NumCatalogPurchases`, `NumStorePurchases`, `NumDealsPurchases` | 채널별 구매 횟수 |
| **고객 특성** | `Income`, `Recency` | 소득 및 최근 구매 경과일 |

## Step 3: 최적의 군집 수 결정

Elbow Method와 Silhouette Score로 최적 *K* 를 결정한다.

- - -
**[참고]**  
군집분석에서 가장 중요한 문제 중 하나는 **군집 개수 K를 몇 개로 설정할 것인가**이다.  
이를 판단하기 위한 대표적인 방법이 Elbow Method와 Silhouette Score이다.

**Elbow Method (엘보 방법)**  
- 핵심 개념
    + *K* 를 증가시키면 데이터는 더 세밀하게 나뉘므로 오차는 감소한다.
    + 하지만 어느 시점 이후에는 *K* 를 증가시켜도 오차 감소 폭이 크지 않다.
    + 이때의 **변화가 꺾이는 지점(Elbow)** 을 최적의 *K* 로 선택한다.
- 동작 원리
    + K = 1, 2, 3, ... 로 증가시키면서
    + 각 *K* 에 대해 군집 내부 오차(WCSS)를 계산한다.
    + *K* 와 오차를 그래프로 나타낸다.
- 그래프는 다음과 같은 형태를 가진다.
```
오차
│
│ ●
│ ●
│ ●
│ ● ← 꺾이는 지점
│ ●
└──────────────
K
```

- 해석
    + 초기에는 *K* 가 증가할수록 오차가 크게 감소한다.
    + 그러나 특정 지점 이후에는 감소 폭이 작아진다.
    + 이때의 꺾이는 지점이 적절한 군집 개수이다.
- 직관적 비유
    + 인력을 추가하면 업무 효율은 증가한다.
    + 하지만 일정 수준 이후에는 사람을 더 추가해도 효율 증가가 미미하다.
    + 이때의 인력 수가 최적 수준과 유사한 개념이다.

**Silhouette Score (실루엣 점수)**
- 핵심 개념
    + 각 데이터가 다음 두 조건을 얼마나 잘 만족하는지를 측정한다.
        * 자신의 군집과는 얼마나 가까운가?
        * 다른 군집과는 얼마나 떨어져 있는가?
    + 이를 하나의 점수로 나타낸다.
- 점수 범위
    + -1 ~ 1 사이 값을 가진다.

| 값 범위 | 의미 |
|--------|------|
| 1에 가까움 | 군집이 잘 분리됨 |
| 0 근처 | 군집 간 경계가 모호함 |
| -1에 가까움 | 잘못된 군집에 속함 |

- 계산 개념
    + 각 데이터에 대해 다음을 계산한다.
        * a: 같은 군집 내 평균 거리
        * b: 가장 가까운 다른 군집까지의 평균 거리
    + 즉, 자기 군집과는 가깝고 다른 군집과는 멀수록 좋은 구조이다.
- 사용 방법
    + 여러 *K* 값에 대해 Silhouette Score를 계산한다.
    + 평균 점수가 가장 높은 *K* 를 선택한다.
- 직관적 비유
    + 한 사람이 속한 팀 내부에서는 서로 잘 어울리고,
    + 다른 팀과는 명확하게 구분되는 상태가 좋은 군집 구조이다.

**두 방법 비교**

| 구분 | Elbow Method | Silhouette Score |
|------|-------------|------------------|
| 기준 | 오차 감소량 | 군집 분리 품질 |
| 선택 방식 | 그래프의 꺾이는 지점 | 점수가 최대인 K |
| 장점 | 직관적이고 계산이 빠름 | 정량적 평가 가능 |
| 단점 | 꺾이는 지점이 모호할 수 있음 | 계산 비용이 더 큼 |

**실무적 활용**
- Elbow Method는 대략적인 *K* 의 범위를 파악하는 데 사용한다.
- Silhouette Score는 그 범위 내에서 최적의 *K* 를 선택하는 데 사용한다.
- 즉,  
    + Elbow Method로 후보를 좁히고,
    + Silhouette Score로 최종 결정을 내리는 방식이 효과적이다.
- - -

## Step 4: K-Means 군집화

결정한 *K* 로 최종 군집화를 실행한다.

- - -
**[참조]**  
**K-Means 군집화 (K-Means Clustering)**
- 핵심 개념
    + K-Means 군집화는 **데이터를 *K* 개의 유사한 그룹(군집)으로 나누는 대표적인 비지도학습 알고리즘**이다.
    + 비슷한 데이터끼리 같은 군집에 속하도록 한다.
- 동작 원리
    1. Step 1: *K* 값 설정
        - 군집 개수 *K* 를 미리 지정한다.
    1. Step 2: 초기 중심점 선택
        - *K* 개의 중심(centroid)을 랜덤으로 설정한다.
    1. Step 3: 군집 할당
        - 각 데이터를 가장 가까운 중심점에 할당한다.
    1. Step 4: 중심 업데이트
        - 각 군집의 평균값으로 중심을 다시 계산한다.
    1. Step 5: 반복
        - 군집이 더 이상 변하지 않을 때까지 반복한다.
- 핵심 특징
    - 거리 기반 알고리즘 (보통 유클리드 거리 사용)
    - 순회(iterative) 방식으로 최적 군집을 찾는다.
    - 초기값에 따라 결과가 달라질 수 있다.
- 장점
    - 구조가 단순하고 이해하기 쉽다.
    - 계산 속도가 빠르다.
    - 대규모 데이터에 적용 가능하다.
- 단점
    - *K* 값을 미리 정해야 한다.
    - 초기 중심값에 민감하다.
    - 구형(원형) 군집에 적합하다.
    - 이상치(outlier)에 영향을 받는다.
- 활용 예시
    - 고객 세분화 (Customer Segmentation)
    - 시장 분석
    - 이미지 압축
    - 이상 패턴 탐지
- - -

````python
user_message = '''앞서 결정한 최적 K로 K-Means 군집화를 실행해 주시기 바랍니다.

1. 최적 K로 K-Means 군집화 실행 (random_state=11)
2. 원본 데이터프레임에 Cluster 컬럼 추가 (0부터 시작하는 정수)
3. 군집별 고객 수와 비율을 표로 제시
4. 군집별 원본 스케일 피처 평균값을 표로 제시
   (표준화 이전 값 기준으로 해석이 용이하도록)
'''
ask(agent, user_message, config)
````

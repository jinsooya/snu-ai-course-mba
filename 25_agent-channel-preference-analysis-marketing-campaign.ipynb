# 채널 선호도 분석이란?

채널 선호도 분석은 고객이 **어떤 구매 채널을 얼마나 자주 이용하는지**를 파악하는 분석 기법이다.

| 채널 | 컬럼명 | 의미 |
|------|--------|------|
| **웹** | `NumWebPurchases` | 웹사이트에서 구매한 횟수 |
| **카탈로그** | `NumCatalogPurchases` | 카탈로그를 통해 구매한 횟수 |
| **매장** | `NumStorePurchases` | 오프라인 매장에서 구매한 횟수 |
| **할인** | `NumDealsPurchases` | 할인 행사를 통해 구매한 횟수 |
| **웹 방문** | `NumWebVisitsMonth` | 한 달간 웹사이트 방문 횟수 |

## 왜 채널 선호도 분석인가?

- **채널별 마케팅 예산 배분**: 어떤 채널에서 구매가 많이 일어나는지 알아야 투자 우선순위를 정할 수 있다.
- **고객 세그먼트별 접점 최적화**: 나이, 소득, 가족 구성에 따라 선호 채널이 다르다.
- **전환율 개선**: 웹 방문 대비 실제 구매 전환율을 채널별로 분석하여 이탈 원인을 파악한다.

## 채널 선호도 분석 방법

채널 선호도는 다음 세 가지 관점으로 분석한다:

| 관점 | 분석 방법 | 질문 |
|------|-----------|------|
| **분포** | 채널별 평균, 중위수, 비율 | '어떤 채널에서 구매가 가장 많이 일어나나?' |
| **특성 연결** | 소득, 나이, 교육별 채널 이용 비교 | '어떤 고객이 어떤 채널을 선호하나?' |
| **전환율** | 웹 방문 대비 웹 구매 비율 | '웹 방문자 중 실제로 구매하는 비율은?' |

## 이 실습에서 사용할 변수

이 실습에서 분석할 채널 관련 컬럼과 고객 특성 컬럼은 다음과 같다.

**채널 컬럼 (5개)**

| 컬럼 | 설명 |
|------|------|
| `NumDealsPurchases` | 할인 행사를 통한 구매 횟수 |
| `NumWebPurchases` | 웹사이트를 통한 구매 횟수 |
| `NumCatalogPurchases` | 카탈로그를 통한 구매 횟수 |
| `NumStorePurchases` | 오프라인 매장에서의 구매 횟수 |
| `NumWebVisitsMonth` | 한 달간 웹사이트 방문 횟수 |

**고객 특성 컬럼**

| 컬럼 | 설명 |
|------|------|
| `Income` | 연 소득 |
| `Year_Birth` | 출생연도 (나이 계산용) |
| `Education` | 교육 수준 |
| `Marital_Status` | 결혼 상태 |
| `Kidhome` / `Teenhome` | 가정 내 어린이/청소년 수 |

```python
import io, logging, sys, platform
from pathlib import Path
from matplotlib import pyplot
from langchain.tools import tool
from utils import pyplot_show_saves_png_first, display_remaining_pyplot_figures, apply_korean_matplotlib_font, resolve_korean_font_family, strip_font_family_overrides, strip_result_folder_reassignment

# --- 출력 폴더 (결과 파일들 저장용)
RESULT_FOLDER = Path('marketing-campaign/channel-result')
RESULT_FOLDER.mkdir(parents=True, exist_ok=True)

# --- 코드 실행용 공유 네임스페이스
_ns = {'__builtins__': __builtins__, 'RESULT_FOLDER': RESULT_FOLDER}
_KOREAN_FONT = resolve_korean_font_family()
exec(f'''
import pandas
import numpy
from matplotlib import pyplot
import seaborn
import warnings
warnings.filterwarnings('ignore')
pyplot.style.use('seaborn-v0_8-whitegrid')
pyplot.rcParams['figure.figsize'] = (10, 6)
pyplot.rcParams['font.size'] = 12
pyplot.rcParams['axes.unicode_minus'] = False
pyplot.rcParams['font.family'] = {_KOREAN_FONT!r}
''', _ns)


@tool(
    'python_execute',
    parse_docstring=True,
    description=(
        'Execute Python code and return the output. '
        'RESULT_FOLDER is pre-set to marketing-campaign/channel-result; do not reassign it. '
        'Use pandas, numpy, matplotlib, seaborn for data analysis and visualization. '
        'Always use print() to display results.'
    )
)
def python_execute(code: str) -> str:
    """
    Execute Python code and return the output.
    Python 코드를 실행하고 결과를 반환한다.

    Args:
        code (str): Python code to execute.
                    실행할 Python 코드

    Returns:
        str: Execution output or error message.
             실행 결과 또는 오류 메시지
    """
    logging.getLogger('matplotlib.font_manager').setLevel(logging.ERROR)
    old_stdout = sys.stdout
    sys.stdout = buffer = io.StringIO()
    try:
        code = strip_font_family_overrides(strip_result_folder_reassignment(code))
        _ns['RESULT_FOLDER'] = RESULT_FOLDER
        with pyplot_show_saves_png_first(_ns.get('pyplot'), RESULT_FOLDER):
            exec(code, _ns)
        _ns['RESULT_FOLDER'] = RESULT_FOLDER
        output = buffer.getvalue()
        display_remaining_pyplot_figures(pyplot, RESULT_FOLDER, apply_font=apply_korean_matplotlib_font)
        return output if output else '실행 완료'
    except Exception as e:
        return f'오류: {type(e).__name__}: {e}'
    finally:
        sys.stdout = old_stdout


# --- 노트북 matplotlib 한글 폰트
apply_korean_matplotlib_font(pyplot)
print(f'OS: {platform.system()}')
print(f'한글 폰트: {pyplot.rcParams['font.family']}')
```

## Step 4: 채널 조합 패턴 분석

여러 채널을 함께 이용하는 패턴(멀티채널)과 단일 채널에만 집중하는 패턴을 분석한다.

- - -
**[참고]**  
**피어슨 상관계수 (Pearson Correlation Coefficient)**
- 핵심 개념: 피어슨 상관계수는 **두 변수 간 선형 관계의 방향과 강도를 -1에서 1 사이 값으로 나타내는 지표**이다.
- 값의 범위: -1 ~ 1 사이 값

| 값 | 의미 |
|----|------|
| 1 | 완전한 양의 상관관계 |
| 0 | 상관관계 없음 |
| -1 | 완전한 음의 상관관계 |

- 해석
    - 값이 1에 가까울수록 -> 두 변수는 함께 증가  
    - 값이 -1에 가까울수록 -> 한 변수 증가 시 다른 변수 감소  
    - 값이 0에 가까울수록 -> 관계 없음  
- 예시
    - 매출 ↑, 광고비 ↑ -> **양의 상관관계 (+)**  
    - 가격 ↑, 수요 ↓ -> **음의 상관관계 (-)**  
    - 키와 신발 사이즈 -> **상관관계 있음**  
    - 키와 좋아하는 색 -> **상관관계 없음**
- 공식
    - 두 변수 X, Y에 대해: `r = Cov(X, Y) / (σ_X * σ_Y)`
        - Cov(X, Y): 공분산  
        - σ_X, σ_Y: 표준편차  
- 특징
    - 선형 관계만 측정 가능  
    - 이상치(outlier)에 민감  
    - 단위에 영향을 받지 않음  
- - -

```python
user_message = '''채널 조합 패턴을 분석하세요.

1. 채널 간 피어슨 상관계수 행렬 계산 (NumDealsPurchases, NumWebPurchases, NumCatalogPurchases, NumStorePurchases)
2. 상관계수 히트맵 시각화 (seaborn heatmap, annot=True, 색상 팔레트: coolwarm)
3. 이용 채널 수(4개 채널 중 구매 횟수 > 0인 채널 수)를 계산하여 분포 출력
   - 0개 채널(아무것도 안 구매) ~ 4개 채널(모든 채널 이용) 고객 수와 비율
4. 멀티채널 고객(3개 이상 채널 이용)의 평균 총구매횟수와 단일채널 고객 비교

분석 결과와 함께 채널 조합 패턴의 인사이트를 한국어로 설명하세요.
'''
ask(agent, user_message, config)
```

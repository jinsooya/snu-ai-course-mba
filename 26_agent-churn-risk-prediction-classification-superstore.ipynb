# 고객 이탈 예측이란?


고객 이탈 예측(Churn Prediction)은 **어떤 고객이 더 이상 구매하지 않을지 미리 예측**하는 분석 기법이다.

| 구분 | 내용 |
|------|------|
| **목적** | 이탈 전에 선제적으로 개입하여 고객을 유지 |
| **방법** | 지도 학습 —> 과거 구매 이력으로 이탈 여부를 학습 |
| **핵심** | 이탈한 고객보다 **이탈 직전 고객**을 잡는 것이 훨씬 효과적 |

## 왜 이탈 예측인가?

- 신규 고객 유치 비용은 기존 고객 유지 비용의 **5~7배**다.
- 이탈 고객의 **5%만 붙잡아도** 수익이 25~95% 증가한다는 연구 결과가 있다.
- 이탈 신호를 조기에 포착하면 **맞춤형 리텐션 캠페인**으로 이탈을 방지할 수 있다.

## 핵심 개념: 이탈 정의

이탈(Churn)은 **데이터와 비즈니스 맥락에 따라 다르게 정의**된다.

| 이탈 정의 방식 | 설명 | 예시 |
|-------------|------|------|
| **시간 기반** | 일정 기간 구매 없으면 이탈 | 최근 6개월 구매 없음 |
| **행동 기반** | 특정 행동 감소 | 방문 횟수 급감 |
| **명시적** | 계정 해지, 구독 취소 | 구독 서비스 해지 |

이 실습에서는 **시간 기반 이탈 정의**를 사용한다.
- 기준일(Reference Date) = 데이터셋 내 최대 주문일
- **최근 6개월 이내 구매 없으면 이탈(1), 있으면 활성(0)**

## 이 실습에서의 피처 및 타겟 변수

이 실습에서 사용할 Superstore 판매 데이터의 변수는 다음과 같다.

| 구분 | 변수 | 설명 |
|------|------|------|
| **타겟** | `Churned` | 이탈 여부 (1=이탈, 0=활성) <— 파생 변수 |
| **Recency** | `recency_days` | 기준일로부터 마지막 구매까지 경과일 |
| **Frequency** | `total_orders` | 총 주문 건수 |
| **Monetary** | `total_sales` | 총 구매 금액 |
| **행동** | `avg_order_value`, `lifespan_days` | 평균 주문 금액, 첫 구매~마지막 구매 기간 |
| **카테고리** | `pct_furniture`, `pct_office_supplies`, `pct_technology` | 카테고리별 구매 비중 |
| **고객 특성** | `Segment` (인코딩) | 고객 세그먼트 |

## Step 3: 학습/테스트 분리 및 클래스 불균형 처리

데이터를 분리하고 클래스 불균형을 처리한다.

- - -
**[참고]**  
**train_test_split**
- 핵심 개념: 데이터를 **학습용(train)** 과 **평가용(test)** 으로 나누는 함수이다.
- 왜 필요한가?
    + 모델이 학습한 데이터를 그대로 평가하면 성능이 과대평가된다.  
    + 따라서 **보지 않은 데이터(test)** 로 평가해야 한다.
- 주요 옵션 설명
    + `test_size=0.2` -> 전체 데이터의 20%를 테스트 데이터로 사용
    + `random_state=11` -> 항상 동일한 방식으로 데이터 분리 (재현성 확보)
    + `stratify=y` -> 타겟 변수 비율을 유지하면서 분리
- 중요한 포인트: 특히 분류 문제에서는 **train과 test의 클래스 비율이 같아야 한다** -> `stratify` 사용

**SMOTE (Synthetic Minority Over-sampling Technique)**
- 핵심 개념: 데이터가 불균형할 때 **적은 클래스(소수 클래스)를 인위적으로 늘리는 방법**이다.
    + 불균형 데이터를 균형 있게 만들어주는 오버샘플링 기법이다.
- 왜 필요한가?
    + 예를 들어:
        * Response = 0 -> 95%
        * Response = 1 -> 5%
    + 이 경우 모델은 그냥 0만 예측해도 높은 정확도를 얻는다. -> 실제로는 좋은 모델이 아니다.
- SMOTE 동작 방식: 단순 복제가 아니라 기존 소수 클래스 데이터를 기반으로 **새로운 데이터를 만들어냄 (합성 데이터)**
- 주요 옵션
    + `random_state=11` -> 결과 재현성 확보

**전체 흐름 정리**
1. Step 1: 데이터 분리
    - train 80%
    - test 20%
    - `stratify=y`로 비율 유지
1. Step 2: 클래스 분포 확인
    - 학습 데이터에서 0과 1의 비율 확인
    - 불균형 여부 파악
1. Step 3: SMOTE 적용
    - 학습 데이터에만 적용 (**중요!!!**)
    - 소수 클래스 데이터 증가
1. Step 4: 적용 전후 비교
    - SMOTE 전 -> 불균형 상태
    - SMOTE 후 -> 균형 상태

**핵심 주의사항 (중요)**
- SMOTE는 **반드시 학습 데이터에만 적용**해야 한다  
- 테스트 데이터에 적용하면 데이터 누수(data leakage)가 발생한다.
- - -

```python
user_message = '''학습/테스트 데이터 분리와 클래스 불균형 처리를 수행해 주시기 바랍니다.

1. 피처(X)와 타겟(y=Churned) 분리
2. train_test_split으로 학습(80%) / 테스트(20%) 분리
   (random_state=11, stratify=y)
3. 학습 데이터의 클래스 분포 확인 (이탈 vs 활성 비율)
4. SMOTE로 학습 데이터 오버샘플링 (random_state=11)
5. SMOTE 적용 전후 클래스 분포를 비교하여 출력하세요.
'''
ask(agent, user_message, config)
```

## Step 4: 모델 학습

랜덤 포레스트 모델을 학습한다.

- - -
**[참고]**: **Random Forest (랜덤 포레스트)**

**핵심 개념**
+ 랜덤 포레스트는 **여러 개의 의사결정나무(Decision Tree)를 만들어서 결과를 종합하는 모델**이다.
+ 즉, 하나의 모델이 아니라 **여러 모델의 집합(앙상블)** 이다.

**왜 사용하는가?**
+ 단일 의사결정나무는
    * 데이터에 과하게 맞추는 문제(과적합)가 발생하기 쉽다.
+ 랜덤 포레스트는
    * 여러 나무의 결과를 평균 또는 투표로 결합하여
    * **과적합을 줄이고 성능을 안정화한다.**

**동작 원리**
1. Step 1: 데이터 샘플링 (bootstrap)
    - 전체 데이터에서 **중복 허용하여 랜덤 샘플 생성**
    - 각 나무는 서로 다른 데이터를 사용
1. Step 2: 랜덤 변수 선택
    - 각 노드에서 모든 변수를 쓰지 않고
    - **일부 변수만 랜덤으로 선택하여 분할**
1. Step 3: 여러 나무 생성
    - 위 과정을 반복하여 여러 개의 트리 생성
1. Step 4: 결과 결합
    - 분류 문제 → **다수결 투표**
    - 회귀 문제 → **평균값**

**직관적 비유**
+ 여러 전문가에게 동시에 의견을 묻는 상황이다.
    * 한 명의 전문가 -> 판단이 틀릴 수 있음  
    * 여러 전문가 -> 의견을 모으면 더 신뢰 가능  
+ 랜덤 포레스트는 **집단 지성** 모델이다.

**주요 특징**
+ 장점
    * 과적합에 강하다
    * 비교적 높은 성능
    * 변수 중요도(feature importance) 제공
    * 데이터 전처리 요구가 적음
+ 단점
    * 모델이 복잡하여 해석이 어려움
    * 계산량이 많음 (트리 수 많을수록)

**주요 하이퍼파라미터**
+ `n_estimators`: 트리 개수
+ `max_depth`: 트리 깊이 제한
+ `max_features`: 각 분할에서 사용할 변수 개수
+ `random_state:`: 결과 재현성  

**언제 사용하는가?**
+ 비선형 관계가 있는 데이터
+ 변수 간 상호작용이 중요한 경우
+ 기본 성능이 좋은 모델이 필요할 때
- - -

```python
user_message = '''랜덤 포레스트(Random Forest) 분류 모델로 이탈을 예측해 주시기 바랍니다.

1. RandomForestClassifier 학습 (n_estimators=100, random_state=11)
2. SMOTE 적용 학습 데이터로 훈련
3. 테스트 데이터로 예측 수행
4. 다음 평가 지표를 모두 출력하세요:
   - Accuracy, Precision, Recall, F1-Score, ROC-AUC Score
5. 이탈 예측에서 Recall이 특히 중요한 이유를 한국어로 설명해 주세요.
   (이탈 고객을 놓치는 것 vs 정상 고객을 이탈로 잘못 분류하는 것의 비용 비교)
'''
ask(agent, user_message, config)
```

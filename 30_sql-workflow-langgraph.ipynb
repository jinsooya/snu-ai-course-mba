## SQL 쿼리 검증 및 교정 도구 노드 함수 정의

# 그래프 흐름 정의

```python
from langgraph.graph import StateGraph, START, END
from langgraph.graph import MessagesState


# --- graph state
graph = StateGraph(MessagesState)

# --- node
graph.add_node('Tool - Table List', table_listing_tool_node)
graph.add_node('Model - Table Selector', table_selection_node)
graph.add_node('Tool - Table Schema Retriever', db_schema_retrieval_tool_node)
graph.add_node('Model - SQL Generator', sql_generation_node)
graph.add_node('Tool - SQL Validator & Corrector', sql_validation_tool_node)
graph.add_node('Tool - SQL Execution', sql_execution_tool_node)
graph.add_node('Model - Answer Generator', answer_generation_node)

# --- edge
graph.add_edge(start_key=START, end_key='Tool - Table List')
graph.add_edge('Tool - Table List', 'Model - Table Selector')
graph.add_edge('Model - Table Selector', 'Tool - Table Schema Retriever')
graph.add_edge('Tool - Table Schema Retriever', 'Model - SQL Generator')
graph.add_edge('Model - SQL Generator', 'Tool - SQL Validator & Corrector')
graph.add_edge('Tool - SQL Execution', 'Model - Answer Generator')
graph.add_edge('Model - Answer Generator', END)

# --- compile
workflow = graph.compile()
```
